```python
# Travel Agent (Goal-Based Agent)
# Name: Alina Arif | Roll No: 23122069 | Section B | Semester 6th
# Subject: Artificial Intelligence | Submitted to: Miss Uzma
#
# Goal: find the shortest route between two cities.
# Algorithm: Dijkstra's (uniform cost search)

import heapq

# ---------- Environment (map of cities, distance in km) ----------
graph = {
    "Karachi":   {"Hyderabad": 165, "Sukkur": 470},
    "Hyderabad": {"Karachi": 165, "Sukkur": 330},
    "Sukkur":    {"Karachi": 470, "Hyderabad": 330, "Multan": 400},
    "Multan":    {"Sukkur": 400, "Lahore": 340, "Islamabad": 550},
    "Lahore":    {"Multan": 340, "Islamabad": 380},
    "Islamabad": {"Multan": 550, "Lahore": 380},
}

COST_PER_KM = 25      # Rs per km (fuel + other expenses)
SPEED_KMH = 70        # average speed of travel


# ---------- Agent function ----------
def travel_agent(start, goal):
    """Returns (path, distance) of the shortest route from start to goal."""
    queue = [(0, start, [start])]     # (distance so far, current city, path)
    visited = set()

    while queue:
        dist, city, path = heapq.heappop(queue)   # take the cheapest option

        if city == goal:                          # goal test
            return path, dist

        if city in visited:
            continue
        visited.add(city)

        for neighbor, d in graph[city].items():   # expand neighbours
            if neighbor not in visited:
                heapq.heappush(queue, (dist + d, neighbor, path + [neighbor]))

    return None, None                             # no route found


# ---------- Percepts (user input) ----------
print("=== Travel Agent ===")
print("Available cities:", ", ".join(graph.keys()))

start = input("Enter start city: ").strip().title()
goal = input("Enter destination city: ").strip().title()

# ---------- Actions (output) ----------
if start not in graph or goal not in graph:
    print("Sorry, city not found. Please choose from the available cities.")
elif start == goal:
    print("You are already at your destination!")
else:
    path, dist = travel_agent(start, goal)
    if path is None:
        print("No route found.")
    else:
        print("\nBest route :", " -> ".join(path))
        print("Distance   :", dist, "km")
        print("Time       : about", round(dist / SPEED_KMH, 1), "hours")
        print("Cost       : about Rs", dist * COST_PER_KM)
```
