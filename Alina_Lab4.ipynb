```python
!pip install -q -U google-genai

from getpass import getpass
from google import genai

API_KEY = getpass("Enter your Gemini API key: ").strip()
client = genai.Client(api_key=API_KEY)
MODEL = "models/gemini-3.6-flash"

university_data = {
    "BSSE Semester 4A": {
        "OOP": "Sir Faisal",
        "DBMS": "Mam Rabia",
        "Data Structures": "Sir Imran",
        "Software Requirements": "Mam Saima"
    }
}


def university_agent(user_input):

    query = user_input.lower()

    if "4a" in query or "semester 4a" in query:

        if "oop" in query or "object oriented" in query:
            return "BSSE Semester 4A - OOP Teacher: Sir Faisal"

        elif "dbms" in query or "database" in query:
            return "BSSE Semester 4A - DBMS Teacher: Mam Rabia"

        elif "data structure" in query:
            return "BSSE Semester 4A - Data Structures Teacher: Sir Imran"

        elif "requirement" in query:
            return "BSSE Semester 4A - Software Requirements Teacher: Mam Saima"

        else:
            return f"""
BSSE Semester 4A Information:

OOP: Sir Faisal
DBMS: Mam Rabia
Data Structures: Sir Imran
Software Requirements: Mam Saima
"""

    prompt = f"""
You are a helpful university information assistant for:

Federal Urdu University of Arts, Science & Technology (FUUAST).

Answer general questions about the university, such as:

- Fees
- Departments
- Campuses
- Admissions
- Programs
- Faculties
- Contact information

For general university information, use your general knowledge.

IMPORTANT:
Do NOT invent specific BSSE Semester 4A teacher information.
That information is handled separately by a Python dictionary.

If information is uncertain or may have changed, clearly mention
that the student should verify it from the university.

Student Question:
{user_input}

Answer clearly and simply:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text


print("FUUAST Student Help Desk Started")
print("Type 'exit' to stop.\n")

while True:

    user = input("Student: ")

    if user.lower() == "exit":
        print("Goodbye!")
        break

    try:
        reply = university_agent(user)

        print("\nHelp Desk:")
        print(reply)
        print()

    except Exception as e:
        print("Error:", e)
```
