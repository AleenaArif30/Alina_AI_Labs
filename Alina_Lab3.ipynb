```python
!pip install -q -U google-genai

from getpass import getpass
from google import genai

API_KEY = getpass("Enter your Gemini API key: ").strip()
client = genai.Client(api_key=API_KEY)

MODEL = "models/gemini-3.6-flash"

def smart_thermostat_agent(user_input):

    prompt = f"""
You are SmartCool AI, a smart thermostat and air conditioner agent.

Task: Read the room conditions given by the user (sensor readings) and decide what the AC should do.

Rules:
1. If temperature is 1 degree or more above the set temperature and the room is occupied, turn the AC ON.
2. If temperature is 1 degree or more below the set temperature, turn the AC OFF.
3. If the room has been empty for 15 minutes or more, switch to ECO mode or turn the AC OFF.
4. If temperature is above 40 degrees C, use MAXIMUM fan speed and show an ALERT.
5. If humidity is above 70 percent, suggest DRY mode.
6. If the set temperature is missing, assume 24 degrees C.

Output format (plain text only, no markdown):
Percept: (short summary of the sensor readings)
Action: (AC ON / AC OFF / ECO MODE)
Fan Speed: (Low / Medium / High / Maximum)
Alert: (message or None)
Reason: (one short sentence)

User sensor readings: {user_input}

Give the decision now:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text

print("SmartCool AI - Smart Thermostat / AC Agent Started (type 'exit' to stop)")
print("Example: temperature 32, humidity 60, room occupied, set temperature 24\n")

while True:
    user = input("Sensor readings: ")

    if user.lower() == "exit":
        print("Goodbye")
        break

    try:
        reply = smart_thermostat_agent(user)
        print(reply)
    except Exception as e:
        print("Error:", e)
```
